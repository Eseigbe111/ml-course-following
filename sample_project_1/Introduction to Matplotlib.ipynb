# 004 Anatomy Of A Matplotlib Figure

### Matplotlib example workflow

```python
# 0) import matplotlib and get it ready for plotting in Jupyter
%matplotlib inline
import matplotlib.pyplot as plt

# 1) Prepare the data u want to plot
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]

# 2) Setup plot
fig, ax = plt.subplots(figsize=(10, 10)) # the figsize (10, 10), is just width and height, 
# which is usually x and y. it is the dimensions of the plot

# 3) Plot data i.e adding data to our axes
ax.plot(x, y)

# 4) Customize plot: This is like naming the plot and their axes
ax.set(title="Simple Plot",
      xlabel="x-axis",
      ylabel="y-axis")

# 5) Save & show (u save the whole figure): U can save it anywhere. But since i have the images 
# folder, i can save them there too.
fig.savefig("images/sample-plot.png")
```

# 005 Scatter Plot And Bar Plot

## Making figures with NumPy arrays:
We want:
* Line plot
* Scatter plot
* Bar plot
* Histogram
* Subplots

```python
import numpy as np
```

```python
# 1) Create sm data
x = np.linspace(0, 10, 100) #(start, stop, num)
x[:10] # we just want to see 10 

# linspace: Return evenly spaced numbers over a specified interval.
# OR
# Returns `num` evenly spaced samples, calculated over the
# interval [`start`, `stop`].
```

```python
# 2) Plot the data and create a line plot (the default in matplotlib is a line plot)
fig, ax = plt.subplots()
ax.plot(x, x**2); # we use ; instead of plt.show()
```

```python
# 3) Use the same data to create a scattered plot
fig, ax = plt.subplots()
ax.scatter(x, np.exp(x));
```

```python
# Another scatter plot
fig, ax = plt.subplots()
ax.scatter(x, np.sin(x))
```

```python
# 4) Making a Bar plot
# Make a plot from a dictionary
nut_butter_prices = {"Almond butter": 10,
                     "Peanut butter": 8,
                     "Cashew butter": 12}
fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax.set(title="Dan's Nut Butter Store",
       ylabel="Price ($)");
```

# 006 Histograms And Subplots

```python
# 5) There are horizontal bars too
fig, ax = plt.subplots()
ax.barh(list(nut_butter_prices.keys()), list(nut_butter_prices.values()))
ax.set(title="Dan's Nut Butter Store",
       ylabel="Price ($)");
```

```python
# 6) Making a Histogram
x = np.random.randn(1000)
fig, ax = plt.subplots()
ax.hist(x);
```

```python
# 7) Making multiple plot happen together i.e a subplot
# Two options for subplot
# a) Option 1
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2,
                                            ncols=2,
                                            figsize=(10, 5))

# Plot to each difft axis
ax1.plot(x, x/2);
ax2.scatter(np.random.random(10), np.random.random(10));
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax4.hist(np.random.randn(1000));
```

# 007 Subplots Option 2

```python
# a) Option 2
fig, ax = plt.subplots(nrows=2,
                       ncols=2,
                       figsize=(10, 5))

# Plot to each difft index
# now, we use difft ax[0, 0] or ax[0, 1] etc so we can see the various plots
ax[0, 0].plot(x, x/2);
ax[0, 1].scatter(np.random.random(10), np.random.random(10));
ax[1, 0].bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax[1, 1].hist(np.random.randn(1000));
```

# 014 Plotting from Pandas DataFrames 5

### Which one should i use? (pyplot vs matplotlib OO Mthd(i.e object oriented mthd))
* when plotting smth quickly, it is okay to use the pyplot mthd
* When plotting smth more advanced, u can use the OO mthd


```python
heart_disease.head()
```

```python
# OO mthd in action
over_50 = heart_disease[heart_disease["age"] > 50]
over_50.head()
```

```python
# 1) pyplot mthd
over_50.plot(kind="scatter", 
            x="age",
            y="chol",
            c="target",
            figsize=(10, 6)); # c is colour, chol is cholesterol
```

```python
# 2) OO mthd mixed with pyplot mthd
fig, ax = plt.subplots(figsize=(10, 6))
over_50.plot(kind="scatter",
            x="age",
            y="chol",
            c="target",
            ax=ax); 
ax.set_xlim([45, 100]); #This changes our x-axis from 45 to 100 instead of 50 to 80 which is default
```

# 015 Plotting from Pandas DataFrames 6

```python
# OO mthd from scratch
fig, ax = plt.subplots(figsize=(10, 6))

# Plot the data
scatter = ax.scatter(x=over_50["age"],
                    y=over_50["chol"],
                    c=over_50["target"]);

# Customize the plot
ax.set(title="Heart Disease and Cholesterol Levels",
      xlabel="Age",
      ylabel="Cholesterol");

# Add a legend
ax.legend(*scatter.legend_elements(), title="Target");

# Add an horizontal line
ax.axhline(over_50["chol"].mean(),
          linestyle="--")
```

# 016 Plotting from Pandas DataFrames 7

##### So now, we are really going to step it up a notch, we are going to see the full power of the matplotlib OO mthd. we're gonna build upon what we've done above. So we're going to add a subplot to this, so we can visualize cholesterol levels and we might get our over 50 data frame from here so we can stop having to scroll up.

```python
# A Subplot of chol, age and thalach
fig, (ax0, ax1) = plt.subplots(nrows=2,
                              ncols=1,
                              figsize=(10, 10));

# 1) THIS IS FOR THE CHOLESTEROL
# Add data to ax0
scatter = ax0.scatter(x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"]);

# Customize ax0
ax0.set(title="Heart Disease and Cholesterol Levels",
       xlabel="Age",
       ylabel="Cholesterol");

# Add a legend to ax0
ax0.legend(*scatter.legend_elements(), title="Target");

# Add a meanline
ax0.axhline(y=over_50["chol"].mean(),
           linestyle="--");


# 2) THIS IS FOR THE THALACH
# Add data to ax1
scatter = ax1.scatter(x=over_50["age"],
                      y=over_50["thalach"],
                      c=over_50["target"]);

# Customize ax1
ax1.set(title="Heart Disease and Max Heart Rate",
       xlabel="Age",
       ylabel="Max Heart Rate");

# Add a legend to ax1
ax1.legend(*scatter.legend_elements(), title="Target");

# Add a meanline
ax1.axhline(y=over_50["thalach"].mean(),
           linestyle="--");

# Add atitle to the figure
fig.suptitle("Heart Disease Analysis", fontsize=16, fontweight="bold");

# NOW FOR THE GRAPHS, U COULD HAVE A SCATTERED PLOT AND A HISTOGRAM OR ANY PLOT, NOT NECESSARILY
# A SECOND SCATTERED PLOT
```

```python
# FOR THE PLOT ABOVE WE CAN REMOVE xlabel value from ax0 and use sharex after figsize=(10, 10), 
# bcos we have the same xlabel axis. This is what i mean below:

# A Subplot of chol, age and thalach
fig, (ax0, ax1) = plt.subplots(nrows=2,
                              ncols=1,
                              figsize=(10, 10),
                              sharex=True);

# 1) THIS IS FOR THE CHOLESTEROL
# Add data to ax0
scatter = ax0.scatter(x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"]);

# Customize ax0
ax0.set(title="Heart Disease and Cholesterol Levels",
       # xlabel="Age",
       ylabel="Cholesterol");

# Add a legend to ax0
ax0.legend(*scatter.legend_elements(), title="Target");

# Add a meanline
ax0.axhline(y=over_50["chol"].mean(),
           linestyle="--");


# 2) THIS IS FOR THE THALACH
# Add data to ax1
scatter = ax1.scatter(x=over_50["age"],
                      y=over_50["thalach"],
                      c=over_50["target"]);

# Customize ax1
ax1.set(title="Heart Disease and Max Heart Rate",
       xlabel="Age",
       ylabel="Max Heart Rate");

# Add a legend to ax1
ax1.legend(*scatter.legend_elements(), title="Target");

# Add a meanline
ax1.axhline(y=over_50["thalach"].mean(),
           linestyle="--");

# Add atitle to the figure
fig.suptitle("Heart Disease Analysis", fontsize=16, fontweight="bold");
```

# 018 Customizing Your Plots 2

```python
# Set the style
plt.style.use('seaborn-v0_8-whitegrid')

# OO mthd from scratch
fig, ax = plt.subplots(figsize=(10, 6))

# Plot the data
scatter = ax.scatter(x=over_50["age"],
                    y=over_50["chol"],
                    c=over_50["target"],
                    cmap="winter"); # This changes the colour scheme

# Customize the plot
ax.set(title="Heart Disease and Cholesterol Levels",
      xlabel="Age",
      ylabel="Cholesterol");

# Add a legend
ax.legend(*scatter.legend_elements(), title="Target");

# Add an horizontal line
ax.axhline(over_50["chol"].mean(),
          linestyle="--");
```

```python
# Customizing the y and x axis limitations
# A Subplot of chol, age and thalach
fig, (ax0, ax1) = plt.subplots(nrows=2,
                              ncols=1,
                              figsize=(10, 10));

# 1) THIS IS FOR THE CHOLESTEROL
# Add data to ax0
scatter = ax0.scatter(x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"],
                      cmap="winter");

# Customize ax0
ax0.set(title="Heart Disease and Cholesterol Levels",
       xlabel="Age",
       ylabel="Cholesterol");

# Change the x axis limits
ax0.set_xlim([50, 80])

# Add a legend to ax0 foe ax0
ax0.legend(*scatter.legend_elements(), title="Target");

# Add a meanline
ax0.axhline(y=over_50["chol"].mean(),
           linestyle="--");


# 2) THIS IS FOR THE THALACH
# Add data to ax1
scatter = ax1.scatter(x=over_50["age"],
                      y=over_50["thalach"],
                      c=over_50["target"],
                      cmap="winter");

# Customize ax1
ax1.set(title="Heart Disease and Max Heart Rate",
       xlabel="Age",
       ylabel="Max Heart Rate");

# Change the x and y axis limits for ax1
ax1.set_xlim([50, 80])
ax1.set_ylim([60, 200])

# Add a legend to ax1
ax1.legend(*scatter.legend_elements(), title="Target");

# Add a meanline
ax1.axhline(y=over_50["thalach"].mean(),
           linestyle="--");

# Add atitle to the figure
fig.suptitle("Heart Disease Analysis", fontsize=16, fontweight="bold");
```
