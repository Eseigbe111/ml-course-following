## 013 Exercise Nut Butter Store Sales

### Dot product example (nut butter sales)

```python
import numpy as np
import pandas as pd
```

```python
#1) Number of jars sold
np.random.seed(0)
product_sold = np.random.randint(20, size=(5, 3))
product_sold
```

```python
#2) Creating a weekly_sales dataframe: i.e putting the above product_sold in a table format
weekly_sales = pd.DataFrame(product_sold, index=["Mon", "Tues", "Wed", "Thurs", "Fri"], 
                            columns=["Almond butter", "Peanut butter", "Cashew butter"])
weekly_sales
```

```python
#3) Create prices array: i.e an array that has the price for each of Almond butter, 
# Peanut butter and Cashew butter, so we can compute the total sold for that week
prices_array = np.array([10, 8, 12])
prices_array
```

```python
# 4) Create butter_prices dataframe: i.e  Almond butter, Peanut butter and Cashew butter in 
# a table format
butter_prices = pd.DataFrame(prices_array.reshape(1, 3), index=["Prices"], columns=["Almond butter", "Peanut butter", "Cashew butter"])
butter_prices
```

```python
product_sold, product_sold.shape
```

```python
prices_array, prices_array.shape
```

```python
# 5) This is just multiplying the arrays to see if it will give corresponding values
total_sales = prices_array.dot(product_sold)
```

```python
total_sales = prices_array.dot(product_sold.T)
total_sales
```

```python
# This is the main calculation of the daily sales
daily_sales = butter_prices.dot(weekly_sales.T)
daily_sales
```

```python
# 6) Now adding it to the weekly_sales column by the right hand side
weekly_sales["Total ($)"] = daily_sales
```

```python
daily_sales
```

```python
# We got error above bcos the shape of daily_sales is (1, 5), but we want it in (5, 1) as the
# column in weekly_sales. So we will need to transpose the daily_sales
weekly_sales["Total ($)"] = daily_sales.T
weekly_sales
```
