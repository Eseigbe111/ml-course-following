### Detremining which sum mthd to use

##### Use Python's mthd on Python datatypes i.e (sum()) and,

##### Use Numpy's mthd on Numpy arrays i.e (np.sum())

```python
massive_array = np.random.random(100000)
massive_array.size
```

```python
massive_array[:10] # Viewing the first 10
```

```python
# %timeit is a magic fc in jupyter notebook or python fc
# This will time the process we perform

%timeit sum(massive_array) # Python 's sum()
%timeit np.sum(massive_array) # Numpy's np.sum()
```

##### From the above results, we can see that Numpy's sum is faster than pythons sum bcos Numpy's sum() is "198.1308 times" faster than Python's sum(). So this means any arithmetic we want to do, we should use the one that is common to that mthd i.e Python's mthd on Python datatypes OR Numpy's mthd on Numpy arrays

# 012 Dot Product vs Element Wise

##### This part tells us when should we use the reshape and transpose mthds

#### Dot Product: This is used when transposing a matrix

```python
np.random.seed(0)
mat1 = np.random.randint(10, size=(5, 3))
mat2 = np.random.randint(10, size=(5, 3))

mat1
```

```python
mat2
```

```python
mat1.shape, mat2.shape
```

#### 1) Element-wise multiplication (Hadamard product)

```python
mat1 * mat2
```

#### 2) Dot product

```python
np.dot(mat1, mat2)
```

```python
# To be able to solve the above, we will need to transpose one of the matrices
mat2_T = mat2.T
mat2_T 
```

```python
mat3 = np.dot(mat1, mat2_T)
mat3
```

# 013 Exercise Nut Butter Store Sales

### Dot product example (nut butter sales)

```python
np.random.seed(0)
# 1) Number of jars sold
sales_amounts = np.random.randint(20, size=(5, 3))
sales_amounts
```

```python
#2) Creating a weekly_sales dataframe
weekly_sales = pd.DataFrame(sales_amounts, index=["Mon", "Tues", "Wed", "Thurs", "Fri"], columns=["Almond butter", "Peanut butter", "Cashew butter"] )
weekly_sales
```

```python
#3) Create prices array
prices = np.array([10, 8, 12])
prices
```

```python
# 4) Create butter_prices dataframe
# The shape of prices is (3,). So we reshape it to (1, 3)
# without prices.reshape(1, 3), we will get an error below
butter_prices = pd.DataFrame(prices.reshape(1, 3), index=["Price"], 
                           columns=["Almond butter", "Peanut butter", "Cashew butter"])
butter_prices
```

```python
sales_amounts, prices
```

```python
total_sales = prices.dot(sales_amounts)
```

```python
sales_amounts.shape, prices.shape
```

```python
# For us to get total_sales, we will need to transpose 
total_sales = prices.dot(sales_amounts.T)
total_sales
```

```python
# So either way we do it i.e whichever we transpose, we will still get the same answer
total_sales = sales_amounts.dot(prices.T)
total_sales
```

```python
# Creating daily sales as a column
butter_prices.shape, weekly_sales.shape
```

```python
weekly_sales.T.shape
```

```python
daily_sales = butter_prices.dot(weekly_sales.T)
daily_sales
```

```python
# Putting the daily_sales as a column by the right side of our dataframe
# This does not work bcos its not the right shape
weekly_sales["Total ($)"] = daily_sales
weekly_sales
```

```python
# Putting the daily_sales as a column by the right side of our dataframe
# This does not work bcos its not the right shape
weekly_sales["Total ($)"] = daily_sales.T
weekly_sales
```
