# SECTION 2 — Data Cleaning

### **Why do we clean the data?**

Imagine you want to predict a car's price.

Your dataset looks like this:

```python
car_sales.head(10)
```

#### The model asks:

**"What is the odometer reading for the Honda at index=7?"**

**Answer:**

**"I don't know."**

**Most scikit-learn models cannot train with missing values.**

**So we must handle them first.**

### **Step 1. Make a copy of the data**

***Never clean the original dataset directly.***

```python
car_sales_filled = car_sales.copy() # We made this incasse of mistake, we can correct using
# the original
car_sales_filled
```

### **Step 2. Fill the "Make" column**

***Check how many are missing.***

```python
car_sales_filled["Make"].isna().sum()
```

**Fill Them**

```python
car_sales_filled["Make"] = car_sales_filled["Make"].fillna("Missing")
car_sales_filled["Make"].head(10)
```

### ***We filled `Make` missing columns with `Missing` bcos `Make` is categorical, and we cannot calculate Average `Toyota and BMW`***

### **Step 3. Fill the "Colour" column**

```python
car_sales_filled["Colour"] = car_sales_filled["Colour"].fillna("Missing")
```

### **Step 4. Fill the Odometer column**

```python
car_sales_filled["Odometer (KM)"].isna().sum()
```

**Fill with the mean.**

```python
car_sales_filled["Odometer (KM)"].mean()
```

```python
car_sales_filled["Odometer (KM)"] = car_sales_filled["Odometer (KM)"].fillna(
    car_sales_filled["Odometer (KM)"].mean()
)
```

**Could we use the median?**

***Yes.***

***In fact, if the data has extreme outliers, median is often better.***

### **Step 5. Fill Doors**

**Look at the values:**

```python
car_sales_filled["Doors"].value_counts()
```

**Most cars have 4 doors.**

```python
car_sales_filled["Doors"].mode()[0] 
```

***So we fill them with the mode():***

```python
car_sales_filled["Doors"] = car_sales_filled["Doors"].fillna(
    car_sales_filled["Doors"].mode()[0]
)
```

**Why mode?**

***Doors is really a discrete category.***

***Average doors: 4.27, doesn't make much practical sense.***

***The most common value is usually a better choice.***

### **Step 6. Handle the target column**

***Look at the missing values.***

```python
car_sales_filled["Price"].isna().sum()
```

**Question**

***Can we fill missing prices with the mean?***
***The answer is: No.***

**Why?**

***Because Price is the target (y).***

***If you invent target values, you're teaching the model with made-up answers.***

***That's a bad practice.***

***Instead, remove those rows.***

```python
car_sales_filled.dropna(subset=["Price"], inplace=True)
```

### **Step 7. Verify everything**

```python
car_sales_filled.isna().sum()
```

**Excellent, Now the data is complete**

### **Step 8. Compare before and after**

**Original:**

```python
car_sales.shape
```

**Filled:**

```python
car_sales_filled.shape
```

***You should notice that only the rows with missing Price were removed.***

**Everything else was preserved.***

### **Step 9. Save the cleaned dataset (Optional)**

***This is good practice.***

```python
car_sales_filled.to_csv(
    "car-sales-clean.csv",
    index=False
)
```

## **A very important lesson**

**Many beginners memorize:**

* ***Mean***
* ***Median***
* ***Mode***

**Instead, ask yourself:  What kind of variable is this?**

* ***Categorical → Often fill with "Missing" or the mode.***
* ***Continuous numerical → Often use the mean or median.***
* ***Target variable (y) → Usually remove rows with missing targets rather than filling them.***

***If you understand why you're choosing a method, you'll be able to make good decisions on new datasets instead of relying on memorized rules.***

# SECTION 4 – Build the Baseline Model

### **What is a Baseline Model?**

#### ***A baseline model is simply the first model you train using its default settings.***

***You are not trying to build the best model yet.***

***You are asking:***

***"If I use this model without trying to improve it, how well does it perform?"***

***This gives us a benchmark to compare future improvements.***

### **Step 1: Import the model**

***Since we're predicting Price, this is a regression problem.***

***We'll use a Random Forest Regressor.***

```python
from sklearn.ensemble import RandomForestRegressor
```

### **Step 2: Create the model**

```python
rf_model = RandomForestRegressor(random_state=42)
```

**Notice we didn't specify things like:**

* ***n_estimators***
* ***max_depth***
* ***max_features***

***We're using the default values.***

***This is for the model to learn patterns in the data***

```python
rf_model.fit(X_train, y_train)
```

# SECTION 5 – Compare Different Regression Models

### **Goal:**
**Train four different models on the same training data and compare them using the same evaluation metrics.**

**The models are:**

* ***Linear Regression***
* ***Decision Tree Regressor***
* ***Random Forest Regressor***
* ***KNN Regressor***

### **Step 1: Import the models**

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor
```

### **Step 2: Create the models**

***Instead of creating them one by one, store them in a dictionary.***

```python
models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42),
    "KNN": KNeighborsRegressor()
}
```

### **Step 3: Create an empty list**

***We'll save each model's results here.***

```python
results = []
```

### **Step 4: Train and evaluate every model**

```python
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

for name, model in models.items():

    # Train the model
    model.fit(X_train, y_train)

    # Make predictions
    y_preds = model.predict(X_test)

    # Evaluate
    r2 = r2_score(y_test, y_preds)
    mae = mean_absolute_error(y_test, y_preds)
    mse = mean_squared_error(y_test, y_preds)
    rmse = np.sqrt(mse)

    # Store results
    results.append({
        "Model": name,
        "R²": r2,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse
    })
```

### **Step 5: Create a comparison table**

```python
comparison = pd.DataFrame(results)
comparison
```

### **Step 6: Sort by R²**

***The one having highest R² will be on top***

```python
comparison.sort_values("R²", ascending=False)
```

### **Step 7: Sort by MAE**

```python
comparison.sort_values("MAE")
```

### **Step 8: Sort by RMSE**

```python
comparison.sort_values("RMSE")
```

# SECTION 6 – Cross Validation

### **What is Cross Validation?**

***Up until now you've done: `train_test_split(test_size=0.2)`***

***which produces one train/test split.***

### **The problem is:**

***What if those 200 test rows happened to be easier (or harder) than average?***

***Your evaluation could be misleading.***

***Cross-validation solves this by creating multiple train/test splits.***


### **Step 1: Import**

```python
from sklearn.model_selection import cross_val_score
```

### **Step 2: Perform Cross Validation**

***We'll use the four models again.***

```python
models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42),
    "KNN": KNeighborsRegressor()
}
```

### **Step 3: Create an empty list**

```python
cv_results = []
```

### **Step 4: Calculate CV R²**

```python
for name, model in models.items():

    scores = cross_val_score(
        model,
        X,
        y,
        cv=5,
        scoring="r2"
    )

    cv_results.append({
        "Model": name,
        "Average CV R²": scores.mean(),
        "Std Dev": scores.std()
    })
```

***Because cross-validation automatically splits the entire dataset into training and testing folds.***

***You don't pass X_train and X_test separately.***

### **Step 5: View the results**

```python
cv_results = pd.DataFrame(cv_results)

cv_results
```

### **How to interpret this**

**Average CV R²**

***This is the average R² across all five folds.***

***Higher is better.***

### **Standard Deviation**

***This tells you how stable the model is.***

***Example:***

**Model A: 0.60, 0.61, 0.59, 0.60, 0.61**

**Std Dev = 0.008 -> Very stable**

**Model B: 0.85, 0.30, 0.77, 0.15, 0.62**

**Std Dev: 0.28 -> Very unstable.**

***A lower standard deviation generally means the model is more consistent across different splits.***

# **SECTION 7 – Hyperparameter Tuning**

### **Since Linear Regression won, you might think we should tune it.***

#### **Here's something important:**

##### ***Linear Regression has almost no hyperparameters worth tuning.***

##### ***So `RandomizedSearchCV` and `GridSearchCV` are generally not useful for a basic LinearRegression.***

**Instead, hyperparameter(`using RandomizedSearchCV and GridSearchCV`) tuning is much more valuable for models like:**

* ***Decision Tree***
* ***Random Forest***
* ***KNN***
* ***SVM***

***So, as a learning exercise, we'll tune Random Forest.***

## **A) RandomizedSearchCV**

### **Step 1: Import**

```python
from sklearn.model_selection import RandomizedSearchCV
```

### **Step 2: Create a parameter grid**

```python
rf_grid = {
    "n_estimators": [100, 200, 500, 1000],
    "max_depth": [None, 5, 10, 20],
    "max_features": ["sqrt", "log2"],
    "min_samples_split": [2, 4, 6],
    "min_samples_leaf": [1, 2, 4]
}
```

### **Step 3: Create the model**

```python
rf = RandomForestRegressor(random_state=42)
```

### **Step 4: Create RandomizedSearchCV**

```python
rs_rf = RandomizedSearchCV(
    estimator=rf, # estimator the model to tune.
    param_distributions=rf_grid, # param_distributions → possible values to try
    n_iter=20,  # n_iter=20 → test 20 random combinations.
    cv=5,  # cv=5 → each combination is evaluated with 5-fold cross-validation.
    verbose=2,  # verbose=2 → show progress.
    random_state=42,  # n_jobs=-1 → use all CPU cores.
    n_jobs=-1  # random_state=42 → reproducible search.
)
```

### **Step 5: Fit**

```python
rs_rf.fit(X_train, y_train)
```

***The above might take a while depending on ur computer***

### **Step 6: Best parameters**

```python
rs_rf.best_params_
```

### **Step 7: Best model**

```python
best_rf = rs_rf.best_estimator_
best_rf
```

### **Step 8: Evaluate the tuned model**

```python
y_preds = best_rf.predict(X_test)

print("R² :", r2_score(y_test, y_preds))
print("MAE:", mean_absolute_error(y_test, y_preds))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_preds)))
```

### **An important lesson**

**One of the biggest lessons in machine learning is:**

***Never assume a more complex model is better. Test it.***

***Sometimes a simple Linear Regression beats a Random Forest.***

***Sometimes a tuned Random Forest beats everything else.***

***The only reliable way to know is to measure and compare, just as you've been doing throughout this project.***

# SECTION 8 - GridSearchCV

### **Why use GridSearchCV after RandomizedSearchCV?**
**Think of it this way.**

***Suppose you're looking for the highest mountain.***
* ***RandomizedSearchCV flies over the country and identifies the highest mountain range.***
* ***GridSearchCV hikes around that mountain range to find the exact highest peak.***

**Since your RandomizedSearchCV found the below as best_params_:**

***{
'n_estimators': 1000,
'min_samples_split': 2,
'min_samples_leaf': 1,
    'max_features': 'sqrt',
    'max_depth': 5
}***

#### ***We'll search around these above values***

### **Step 1: Import GridSearchCV**

```python
from sklearn.model_selection import GridSearchCV
```

### **Step 2: Create a smaller parameter grid**
***Notice we don't search huge ranges again.***


```python
grid = {
    "n_estimators": [800, 1000, 1200],
    "max_depth": [4, 5, 6],
    "max_features": ["sqrt"],
    "min_samples_split": [2, 3],
    "min_samples_leaf": [1, 2]
}
```

**Let's calculate how many combinations this is:**

* ***n_estimators → 3***
* ***max_depth → 3*** 
* ***max_features → 1***
* ***min_samples_split → 2***
* ***min_samples_leaf → 2***

**Total:**

***3 × 3 × 1 × 2 × 2 = 36 combinations***

**With cv=5:**

***36 × 5 = 180 model fits***

**This is manageable.**

### **Step 3: Create the model**

```python
rf = RandomForestRegressor(random_state=42)
```

### **Step 4: Create GridSearchCV**

```python
gs_rf = GridSearchCV(
    estimator=rf,
    param_grid=grid,
    cv=5,
    verbose=2,
    n_jobs=-1
)
```

### **Step 5: Fit**

```python
gs_rf.fit(X_train, y_train)
```

***This will take a few minutes.This will take a few minutes.***

### **Step 6: Best parameters**

```python
gs_rf.best_params_
```

### **Step 7: Best estimator**

```python
best_grid_rf = gs_rf.best_estimator_
```

### **Step 8: Predict**

```python
y_preds = best_grid_rf.predict(X_test)
```

### **Step 9: Evaluate**

```python
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

r2 = r2_score(y_test, y_preds)
mae = mean_absolute_error(y_test, y_preds)
mse = mean_squared_error(y_test, y_preds)
rmse = np.sqrt(mse)

print(f"R²   : {r2:.3f}")
print(f"MAE  : {mae:.2f}")
print(f"MSE  : {mse:.2f}")
print(f"RMSE : {rmse:.2f}")
```
