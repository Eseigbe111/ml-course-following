## 012 Manipulating Data 2

```python
# Now how do we create data from existing data?
# Well panda has a few difft ways that u can create data, such as building a new ccolumn. 
# Let's have a look at a couple of difft ways.

# 1) Column from series
seats_column = pd.Series([5, 5, 5, 5, 5])
# The below creates a new column called seats
car_sales["Seats"] = seats_column
car_sales
```

```python
# We want to fill the places having no value in our Seats column
# car_sales["Seats"] = car_sales["Seats"].fillna(5, inplace=True) : While inplace was working 
# well for dave, it did not for me bcos currently, Such inplace method never works to update 
# the original DataFrame or Series, because the intermediate object on which we are setting 
# values always behaves as a copy (due to Copy-on-Write). Instead try using 
# 'df.method({col: value}, inplace=True)' instead, to perform the operation inplace on the 
# original object, or try to avoid an inplace operation using 'df[col] = df[col].method(value)'

car_sales["Seats"] = car_sales["Seats"].fillna(5)
```

```python
car_sales
```

```python
# 2) Column from Python list: When creating a column from a list, it has to be the same length 
#  as ur dataframe, while in series it can be any length and then u fill it later on.
fuel_economy = [7.5, 9.2, 5.0, 9.6, 8.7, 4.7, 7.6, 8.7, 3.0, 4.5]
car_sales["Fuel per 100KM"] = fuel_economy
car_sales
```

```python
car_sales["Total fuel used (L)"] = car_sales["Odometer (KM)"]/100 * car_sales["Fuel per 100KM"]
car_sales
```

```python
# 3) We can also create a column from a single value
car_sales["Number of wheels"] = 4
car_sales 
```

```python
car_sales["Passed road safety"] = True
car_sales
```

```python
# 4) Now to remove a column, we do:
car_sales = car_sales.drop("Total fuel used", axis=1)
car_sales
```
