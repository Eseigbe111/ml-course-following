# **UNDERSTANDING CORRELATION MATRIX**

### **Correlation simply asks:**

> ### ***"When one variable changes, does the other variable tend to change with it?"***

### **There are three basic possibilities:**
> ### ***Positive correlation***

### **When one goes up, the other tends to go up.**

> ### ***Negative correlation**

### **When one goes up, the other tends to go down.**

> ### ***Near zero***

### ***There isn't much of a straight-line relationship between them.***
### **This is what u should know:**
* **+1 → very strong positive relationship**
* **+0.5 → moderate positive relationship**
* **0 → no linear relationship**
* **-0.5 → moderate negative relationship**
* **-1 → very strong negative relationship**

### **The most important thing for you right now is:**
> ### ***Find target and look across that row.***
### **Why?**
> ### ***Because target is what you're trying to predict.***

### **Let's go thru examples**
### **Look at this:**
### **`cp → target = 0.43`**
### **What does that mean?**
### **It means:**
> ### ***There is a `positive relationship` between the encoded cp variable and the encoded target variable in this dataset.***
### **Why positive?**
### **Because the number is +0.43.**
### **Why do we care?**
> ### ***Because it suggests that cp contains some information related to the target.***
### **It does NOT mean:**
> ### ***cp causes heart disease.***
### **It does NOT mean:**
>### ***If cp increases by 1, heart disease increases by 0.43.***

### **So when reading a correlation number, ask two questions:**

### **Question 1: What is the sign?**
* **+ → positive relationship**
* **- → negative relationship**

### **Question 2: How far is it from zero?**
* **Closer to 1 or -1 → stronger relationship.**
* **Closer to 0 → weaker relationship.**

### ***3) Choosing the Right Models***
* **LogisticRegression**

* **RandomForestClassifier**

* **GradientBoostingClassifier**

#### ***Put models in a dictionary, so we do not repeat the same code for all models***

```python
# models dictionary
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "KNN": KNeighborsClassifier(),
    "Random Forest": RandomForestClassifier(),
    "SVC": SVC(),
    "Gradient Boosting": GradientBoostingClassifier()
}

# Create a fc to fit and score models
def fit_and_score(models, X_train, X_test, y_train, y_test):
    """
    Fits and evaluates given ML models.
    models: a dict of difft Scikit-Learn ML models
    X_train: training data (no labels)
    X_test: testing data (No labels)
    y_train: training labels
    y_test: test labels
    """
    # Set random seed
    np.random.seed(42)
    # Make dictionary to keep model scores
    model_scores = {}
    # Loop thru models
    for name, model in models.items():
        # Fit the model to the training data
        model.fit(X_train, y_train)
        # Evaluate the model and append its score to model_scores
        model_scores[name] = model.score(X_test, y_test)
    return model_scores
```

```python
model_scores = fit_and_score(models=models,
                            X_train=X_train,
                            X_test=X_test,
                            y_train=y_train,
                            y_test=y_test)
model_scores
```

## **Class 1 — Feature Engineering**
### ***1) What is Feature engineering***
#### Feature engineering is the process of creating, changing, or selecting features from existing data so that a machine learning model can learn from the data more effectively.

#### A feature is simply an input variable that the model uses to make a prediction.

#### **For example, in your heart-disease dataset:**
> ##### `age, sex, cp, trestbps, chol, thalach, oldpeak,` ... etc are features
> ##### Your target is `target`, which u are trying to predict

#### **Simple idea**
##### **Suppose we have:**
> * ##### height = 1.8 m
> * ##### weight = 90 kg

#### **Instead of giving the model only height and weight, we can create:**
> ##### BMI = weight / height²
##### Now we have created a new feature from existing features.
##### That is feature engineering.

### ***2) Why Do We Need Feature Engineering?***
#### Raw data is not always in the most useful form for a machine learning model.
#### Sometimes the information needed by the model is already present in the dataset, but it is hidden inside other features.

#### **For example:**
> * ##### height = 1.8
> * ##### weight = 90

#### The model could potentially learn the relationship between height and weight itself.
> * ##### But giving it:
> * ##### BMI = 27.78
#### can make an important relationship more directly available.

#### **So feature engineering can help a model:**
> * ##### discover useful relationships
> * ##### represent information in a better form
> * ##### improve prediction performance
> * ##### reduce unnecessary complexity
> * ##### make patterns easier for some models to learn

#### However, feature engineering does not automatically improve a model.
#### A badly created feature can make the model worse.

### ***3) An Important Distinction***

#### **There are three things we can do with features:**

#### **A. Create new features**

#### **Example:**
> ##### df["BMI"] = df["weight"] / (df["height"] ** 2)
#### We created a new feature.

#### **B. Transform existing features**
#### **Example:**
> ##### df["log_income"] = np.log1p(df["income"])
#### We transformed income into another representation.

#### **C. Remove features**
#### Sometimes a feature is:
> * ##### irrelevant
> * ##### duplicated
> * ##### extremely noisy
> * ##### leaking information from the future
> * ##### unnecessary
#### So feature engineering can also involve deciding which features should remain in the model.

#### ***4) Example Using Your Heart-Disease Dataset***
#### **Your dataset contains:**
> ##### age, sex, cp, trestbps, chol, fbs, restecg, thalach, exang, oldpeak, slope, ca, thal, target
#### Let's think about the information.
> ##### age
#### Age tells us how old the patient is.
#### We could potentially create:
> ##### df["age_squared"] = df["age"] ** 2
#### **Why?**
#### **Because perhaps the relationship between age and heart disease isn't simply:**
> ##### Every additional year has exactly the same effect.
#### An age_squared feature allows a model to represent a curved/non-linear relationship more easily.
#### But this does not mean age_squared will necessarily improve our model.
#### We would test it.

#### ***5) Creating a Ratio***
#### Another common type of feature engineering is creating a ratio.
#### **Suppose we have:**
> * ##### income = 100,000
> * ##### household_size = 5
#### **We could create:**
> ##### income_per_person = 100,000 / 5
                  = 20,000
#### The original income tells us how much money the household has.
#### But income per person tells us something different.
#### **Code:**
> ##### df["income_per_person"] = df["income"] / df["household_size"]
#### This is useful because sometimes the relationship between two variables contains more information than either variable individually.

#### ***6) Feature Engineering Doesn't Mean "Create Features Randomly***
#### This is very important.
#### **We should not just create:**
> ##### age_squared, age_cubed, age_log, age_times_chol, age_divided_by_chol, age_plus_chol...
#### just because we can.
#### **Instead, we should ask:**
> ##### Does this new feature represent something meaningful about the problem?
#### **Feature engineering should be guided by:**
> * ##### the problem
> * ##### the meaning of the variables
> * ##### domain knowledge
> * ##### what information is available at prediction time

#### ***7) Data Leakage — VERY IMPORTANT***
#### One of the most important things you need to understand about feature engineering is data leakage.
#### Data leakage happens when information that should not be available to the model at prediction time gets into the training data.
#### Imagine we want to predict whether a patient has heart disease before diagnosis.
#### **Suppose we create a feature using:**
> ##### final_diagnosis
#### But that diagnosis is only known after the doctor has examined the patient.
#### We would be giving the model information that it would not actually have when making the prediction.
#### The model may appear extremely accurate, but the result would be misleading.

#### ***8) A Simple Example of Leakage***
#### Imagine we want to predict whether a student will pass an exam.
#### **Available before the exam:**
> ##### attendance, study_hours, previous_scores
#### These are legitimate features.
#### **But suppose we create:**
> ##### final_exam_score
#### and use it to predict whether the student passed.
#### That's leakage.
#### **Why?**
#### **Because:**
> ##### final_exam_score
#### is only known after the exam.
#### The model wouldn't have that information when the prediction is supposed to be made.
#### **The rule to remember:**
> ##### A feature must only use information that would genuinely be available at the time the prediction is made.

#### ***9) Feature Engineering on Your Heart-Disease Dataset***
#### Let's look at your actual features.
#### **Some possible engineered features could be:**
#### **a) Age transformation**

#### **b) Maximum heart-rate relationship**
#### **You have:**

#### **We could explore a feature such as:**

#### But we have to be careful here.
#### A mathematically possible feature isn't automatically a medically meaningful feature.
#### **So before creating it, we'd ask:**
> * ##### Does this ratio make sense for the medical problem?
#### This is where `domain knowledge` becomes important.

#### ***10) Feature Engineering Can Also Be Binning***
#### Sometimes we convert a continuous variable into groups.
#### **For example:**
> ##### age: 25 31 42 56 67 72
#### **could become:**
> ##### Young, Young, Middle, Middle, Older, Older
#### **For example:**
> ##### df["age_group"] = pd.cut(df["age"], bins=[0, 40, 60, 100], labels=["Young", "Middle", "Older"])
#### Now we've created a categorical feature.
#### But again, we shouldn't automatically assume this is better.
#### Turning a continuous variable into groups can also throw away information.
#### **For example:**
> * ##### age = 59
> * ##### age = 40
#### might end up in different groups even though they aren't dramatically different.
#### So binning should have a good reason.

#### ***11) Feature Engineering vs Feature Selection***
#### These two are related but different.

#### **A) Feature engineering**
#### We create or transform features.
#### **Example:**
> * ##### height + weight → BMI
> * ##### Feature selection
#### We decide which existing features should be used.
#### **Example:**
> ##### age, sex, chol, thalach
#### might be selected while another feature is removed.
#### **So:**
#### Feature engineering = creating better representations of information.

#### **B) Feature selection = deciding which information to keep.**

## **Class 2A — Cross-Validation**
### ***1) What is Cross-Validation?***
#### Cross-validation is a method of evaluating a machine learning model by splitting the training data into several parts and using different parts for validation.
#### It helps us get a more reliable idea of how well a model performs.

### ***2) How does 5-Fold Cross-Validation work?***
#### **With cv=5:**
> ##### the training data is divided into 5 folds.
#### **The model is trained and validated 5 times:**
> * ##### Round 1 → Fold 1 = validation, others = training
> * ##### Round 2 → Fold 2 = validation, others = training
> * ##### Round 3 → Fold 3 = validation, others = training
> * ##### Round 4 → Fold 4 = validation, others = training
> * ##### Round 5 → Fold 5 = validation, others = training
#### Each fold gets a chance to be the validation set.
#### The scores are then averaged.
#### **Example:**
> ##### 0.80, 0.85, 0.90, 0.85, 0.80
#### **Average:**
> ##### 0.84
#### So the cross-validation score is 84%.

### ***3) Why do we use Cross-Validation?***
#### A single train/validation split can give us a score that depends heavily on which observations happened to be selected.
#### Cross-validation gives us multiple validation results, making our evaluation more reliable.

### ***4) Cross-Validation vs Test Set***
### **This is very important:**
> ##### Training data -> Cross-validation -> Choose/tune model -> Final model -> Test data -> Final evaluation

#### The test set should remain untouched until the final evaluation.
#### **Remember:**
> ##### Cross-validation helps us choose and evaluate models during development. The test set gives us the final evaluation.

### ***5) best_score_ vs Test Score***
#### **When you use:**
> ##### gs_knn.best_score_
#### this is the average cross-validation score for the best parameters found during GridSearchCV.
#### **But:**
> ##### gs_knn.score(X_test, y_test)
#### is the model's performance on the untouched test set.
#### They don't have to be the same.

### ***6) Stratified Cross-Validation***
#### For classification problems like your heart-disease project, we generally want each fold to maintain a similar proportion of class 0 and class 1.
#### This is called stratification.
#### **For example:**
#### **Full data:**
> * ##### 60% class 0
> * ##### 40% class 1
#### **Each fold tries to maintain approximately:**
> * ##### 60% class 0
> * ##### 40% class 1
#### Scikit-learn commonly uses stratified folds automatically when cv=5 is used with classification models.

### ***7) When Should We Be Careful?***
#### Different problems can require different cross-validation methods.
#### **For ordinary classification:**
> ##### StratifiedKFold
#### **For time-dependent data such as electricity demand:**
> ##### TimeSeriesSplit
#### because we don't want future information influencing the training of a model that is supposed to predict the future.

## **Key Notes to Remember**
#### **Write these four:**
> * #### Cross-validation = repeatedly splitting training data into training and validation portions.
> * #### cv=5 = 5-fold cross-validation.
> * #### Cross-validation is used during model development; the test set is kept for final evaluation.
> * #### For classification, stratification helps maintain class proportions in each fold

# **Class 3A — Pipelines**
### ***1) What is a Pipeline?***
#### A Pipeline connects preprocessing steps and the machine-learning model into one process.
#### **For example:**
> ##### Scale data → Train KNN
#### Instead of doing these separately, we put them together.

### ***2) Why use a Pipeline?***
#### The main reason is to prevent data leakage during cross-validation.
#### **For example, with KNN:**
> * ##### from sklearn.pipeline import Pipeline
> * ##### from sklearn.preprocessing import StandardScaler
> * ##### from sklearn.neighbors import KNeighborsClassifier
> * ##### knn_pipe = Pipeline([("scaler", StandardScaler()), ("knn", KNeighborsClassifier())])

#### **Then:**
> * ##### knn_pipe.fit(X_train, y_train)
> * ##### knn_pipe.score(X_test, y_test)
#### **The pipeline automatically:**
> * ##### Scales the data.
> * ##### Trains KNN.
> * ##### Uses the same scaling process when making predictions.

### ***3) Important idea***
#### **Without a pipeline:**
> ##### Scale → Model
#### **With a pipeline:**
> ##### Scale → Model → Prediction
#### And during cross-validation, the scaler is fitted only on each training fold, which helps prevent leakage.

## **Key note**
> ##### A Pipeline combines preprocessing and a model into one workflow and helps prevent data leakage during cross-validation.

## **Short classwork**
#### **Create a Logistic Regression pipeline using:**
> * ##### StandardScaler()
> * ##### LogisticRegression(max_iter=2000)
#### **Then run:**
> * ##### pipeline.fit(X_train, y_train)
> * ##### pipeline.score(X_test, y_test)

## **This is the code below:**

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression


# pipeline here is for the model
pipeline = Pipeline([
    ("scaler", StandardScaler()), 
    ("log_reg", LogisticRegression(max_iter=2000))
])

pipeline.fit(X_train, y_train)
```

```python
pipeline.score(X_test, y_test)
```

# **Class 3B — Pipeline + GridSearchCV**
#### The main reason for combining them is that GridSearchCV can tune the preprocessing and model together without data leakage.
#### **Example: KNN**
> * ##### from sklearn.pipeline import Pipeline

> * ##### grid.fit(X_train, y_train)

#### **Notice:**
> ##### "knn__n_neighbors"
#### **The knn__ means:**
> ##### Go inside the Pipeline step named "knn" and tune its parameter.
#### **Check the result**
> * ##### grid.best_params_
> * ##### grid.best_score_
#### **And finally:**
> ##### grid.score(X_test, y_test)

## **This is the code below:**

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV

# Scale using pipeline
knn_pipe = Pipeline([ 
    ("scaler", StandardScaler()), 
    ("knn", KNeighborsClassifier())
])

# Parameters we want to search for
knn_grid = {
    "knn__n_neighbors": [3, 5, 7, 9], 
    "knn__weights": ["uniform", "distance"], 
    "knn__p": [1, 2]
}

# Gridsearchcv
grid = GridSearchCV(
    knn_pipe, 
    knn_grid, 
    cv=5, 
    scoring="accuracy"
)

# Fitting data to training set
grid.fit(X_train, y_train)
```
